```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

```python
N=50
x = np.zeros(N)
x[0] = 1

plt.figure(figsize=(10, 4))
plt.stem(x)
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.title("Discrete-Time Impulse")
plt.grid(True)
plt.show()
```

```python
h = np.ones(5) / 5
plt.figure(figsize=(10, 4))
plt.stem(h)
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.title("5-Sample Moving Average Impulse Response")
plt.grid(True)
plt.show()
```

```python
y = np.convolve(x, h)

plt.figure(figsize=(10, 4))
plt.stem(y)
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.title("Output of Moving-Average Filter to an Impulse")
plt.grid(True)
plt.show()
```

```python
x = np.zeros(50)
x[10:20] = 1
```

```python
y = np.convolve(x, h)

plt.figure(figsize=(10, 4))
plt.plot(x, label="Input")
plt.plot(y, label="Filtered")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.title("Moving-Average Filtering")
plt.legend()
plt.grid(True)
plt.show()
```

```python
w, H = signal.freqz(h, worN=1024, fs=1000)

magnitude_db = 20 * np.log10(np.abs(H))

plt.figure(figsize=(10, 4))
plt.plot(w, magnitude_db)
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB)")
plt.title("Frequency Response of 5-Sample Moving Average")
plt.grid(True)
plt.show()
```

```python
N = 30
a = 0.8

x = np.zeros(N)
x[0] = 1

y = np.zeros(N)

for n in range(N):
    if n == 0:
        y[n] = x[n]
    else:
        y[n] = x[n] + a * y[n - 1]

plt.figure(figsize=(10, 4))
plt.stem(y)
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.title("Impulse Response of a First-Order IIR Filter")
plt.grid(True)
plt.show()
```

```python
fs = 1000
f0 = 100

w0 = 2 * np.pi * f0 / fs
r = 0.8

b = [1]
a = [1, -2 * r * np.cos(w0), r**2]

w, H = signal.freqz(b, a, fs=fs)

plt.figure(figsize=(10, 4))
plt.plot(w, 20 * np.log10(np.abs(H)))
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB)")
plt.title("IIR Resonator")
plt.grid(True)
plt.show()
```

```python
#FIR moving avg
h_fir = np.ones(5) / 5

# IIR first order
a1=0.8

h_iir = np.zeros(N)
h_iir[0] = 1

for n in range(1, N):
    h_iir[n] = a1 * h_iir[n-1]
```

```python
plt.figure(figsize=(10, 4))
plt.stem(h_fir)
plt.title("FIR Impulse Response")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid()
plt.show()

plt.figure(figsize=(10, 4))
plt.stem(h_iir)
plt.title("IIR Impulse Response")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```
